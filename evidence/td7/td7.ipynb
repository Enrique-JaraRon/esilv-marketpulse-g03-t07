# TD07 — Normalisation et comparaison
**Mathis2103 — G03 — T07**

## Objectif
Comparer AAPL au SP500 sur des dates communes à partir des rendements,
des bases 100 et de la performance relative.

## Travail réalisé
- **Mathis** : alignement des séries, performance relative et intégration dans `main.py`.
- **Enrique** : indexation par date, rendement et base 100.

Les cinq fonctions sont regroupées dans `src/analytics.py`.
Les affichages restent dans `main.py`.
Nous avons travaillé sur `feature/analytics`, avec des commits individuels.

## Résultats
L’exécution de `python src/main.py` donne :
- 21 séances communes.
- Rendement AAPL : **+6,48 %**.
- Rendement SP500 : **+2,15 %**.
- Surperformance d’AAPL : **+4,33 points de pourcentage**.
- Bases 100 finales : **106,48** pour AAPL et **102,15** pour SP500.

Ces résultats correspondent aux valeurs attendues du TD.
Une PR commune regroupe nos contributions pour leur intégration dans `main`.
